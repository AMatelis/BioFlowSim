```python
import matplotlib.pyplot as plt
from BioFlowSim.simulation import FlowSimulator
from BioFlowSim.imaging import ArtifactPipeline, GaussianNoiseArtifact, PoissonNoiseArtifact

sim = FlowSimulator(width=100, height=60, num_particles=50, velocity=2.0)
frames = sim.simulate(num_frames=10)

pipeline = ArtifactPipeline(artifacts=[GaussianNoiseArtifact(), PoissonNoiseArtifact()], seed=42)
processed_frames = pipeline.apply_batch(frames)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for i, ax in enumerate([0, 4, 9]):
    ax.imshow(processed_frames[i], cmap='gray')
    ax.set_title(f"Frame {i+1}")
    ax.axis('off')

plt.show()
```
